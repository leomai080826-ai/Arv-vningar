## Del 1: Exempel
Varje exempel är fristående (definierar sina egna klasser från grunden).

### Exempel 1 — Grundläggande arv

Med **arv** kan en klass (subklass) återanvända attribut och metoder från en annan klass (superklass).

Syntaxen är `class Subklass(Superklass):`. Om subklassen inte har något eget innehåll skriver man `pass`.

Här ärver `Hund` från `Djur` utan att ändra något — och får ändå `__init__` och `namn` helt gratis.

```python
class Djur:
    def __init__(self, namn):
        self.namn = namn

class Hund(Djur):   # Hund ärver från Djur
    pass

nero = Hund("Nero")   # __init__ kommer från Djur
print(nero.namn)
```

### Exempel 2 — Ärvda metoder fungerar automatiskt

En subklass får inte bara attribut, utan även metoder från superklassen — helt utan att man behöver skriva om dem.

```python
class Djur:
    def __init__(self, namn):
        self.namn = namn

    def ljud(self):
        print(self.namn + " gör ett ljud.")

class Hund(Djur):
    pass

nero = Hund("Nero")
nero.ljud()   # ljud() är ärvd från Djur
```

### Exempel 3 — Override: skriva över en metod

Om subklassen definierar en metod med **samma namn** som en metod i superklassen, används subklassens version istället. Det kallas att man **overridar** metoden.

```python
class Djur:
    def __init__(self, namn):
        self.namn = namn

    def ljud(self):
        print(self.namn + " gör ett ljud.")

class Hund(Djur):
    def ljud(self):              # skriver över Djurs ljud()
        print(self.namn + " säger: Voff!")

nero = Hund("Nero")
nero.ljud()   # Nero säger: Voff!
```

### Exempel 4 — `super().__init__()`

Om subklassen behöver egna attribut UTÖVER de ärvda, kan man använda `super().__init__(...)` för att köra superklassens konstruktor, och sedan lägga till sina egna rader.

```python
class Djur:
    def __init__(self, namn):
        self.namn = namn

class Hund(Djur):
    def __init__(self, namn, ras):
        super().__init__(namn)   # kör Djurs __init__
        self.ras = ras            # eget, nytt attribut

nero = Hund("Nero", "Labrador")
print(nero.namn, nero.ras)
```

### Exempel 5 — Nya metoder som bara finns i subklassen

Subklassen kan även ha metoder som superklassen inte alls har. Ett rent `Djur`-objekt kan inte hämta boll — bara `Hund`-objekt kan det.

```python
class Djur:
    def __init__(self, namn):
        self.namn = namn

class Hund(Djur):
    def __init__(self, namn, ras):
        super().__init__(namn)
        self.ras = ras

    def hamta_boll(self):
        print(self.namn + " hämtar bollen!")

nero = Hund("Nero", "Labrador")
nero.hamta_boll()
```

### Exempel 6 — Flera subklasser och polymorfism

Man kan ha flera subklasser till samma superklass. Om alla overridar samma metodnamn kan man loopa igenom en lista med blandade objekt och anropa samma metodnamn — varje objekt svarar på sitt eget sätt. Det kallas **polymorfism**.

```python
class Djur:
    def __init__(self, namn):
        self.namn = namn

    def ljud(self):
        print(self.namn + " gör ett ljud.")

class Hund(Djur):
    def ljud(self):
        print(self.namn + " säger: Voff!")

class Katt(Djur):
    def ljud(self):
        print(self.namn + " säger: Miau!")

djur_lista = [Hund("Nero"), Katt("Molly"), Djur("Något djur")]

for d in djur_lista:
    d.ljud()
```

### Exempel 7 — `isinstance()`

`isinstance(objekt, Klass)` kollar om ett objekt tillhör en viss klass — eller en klass som objektets klass ärver från. Ett `Hund`-objekt räknas alltså SOM ett `Djur` också.

```python
class Djur:
    def __init__(self, namn):
        self.namn = namn

class Hund(Djur):
    pass

class Katt(Djur):
    pass

nero = Hund("Nero")

print(isinstance(nero, Hund))   # True
print(isinstance(nero, Djur))   # True — Hund ÄR ETT Djur
print(isinstance(nero, Katt))   # False
```

### Exempel 8 — Anropa superklassens metod med `super()`

`super()` kan också användas för att anropa superklassens version av VILKEN metod som helst, inte bara `__init__`. Här bygger `Hund` vidare på `Djur`s `beskrivning()` istället för att skriva om allt från grunden.

```python
class Djur:
    def __init__(self, namn):
        self.namn = namn

    def beskrivning(self):
        print(self.namn + " är ett djur.")

class Hund(Djur):
    def beskrivning(self):
        super().beskrivning()          # kör Djurs beskrivning() först
        print("Dessutom är det en hund.")

nero = Hund("Nero")
nero.beskrivning()
```

### Exempel 9 — Klassattribut ärvs också

Klassattribut (som delas av alla objekt) fungerar likadant över arv — en subklass kommer åt och kan uppdatera samma klassattribut som superklassen har.

```python
class Djur:
    antal_djur = 0   # klassattribut

    def __init__(self, namn):
        self.namn = namn
        Djur.antal_djur += 1

class Hund(Djur):
    pass

d1 = Djur("Något djur")
h1 = Hund("Nero")
h2 = Hund("Buster")

print("Antal djur totalt:", Djur.antal_djur)
```

### Exempel 10 — Flernivå-arv

En subklass kan i sin tur vara superklass åt en annan klass — en hel kedja av arv. `Valp` ärver från `Hund`, som i sin tur ärver från `Djur`. `Valp` får då med sig allt från BÅDA.

```python
class Djur:
    def __init__(self, namn):
        self.namn = namn

    def ljud(self):
        print(self.namn + " gör ett ljud.")

class Hund(Djur):
    def __init__(self, namn, ras):
        super().__init__(namn)
        self.ras = ras

    def ljud(self):
        print(self.namn + " säger: Voff!")

    def hamta_boll(self):
        print(self.namn + " hämtar bollen!")

class Valp(Hund):
    pass

buster = Valp("Buster", "Beagle")

buster.ljud()          # ärvd via Hund
buster.hamta_boll()    # ärvd via Hund
print(buster.namn)      # ärvd via Djur (genom Hund)
```
